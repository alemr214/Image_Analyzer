## Data

Reading the images from the images folder as arrays from NumPy.

```python
img_mountain = np.asarray(Image.open("../images/mountain.jpg"))
img_flowers = np.asarray(Image.open("../images/flowers.jpg"))
```

---

## Representation

Viewing how an image is represented with NumPy arrays.

```python
print(f"Mountain image shape: {img_mountain.shape} and data type: {img_mountain.dtype}")
print(f"Flowers image shape: {img_flowers.shape} and data type: {img_flowers.dtype}")
```

The shape constitutes three values: height, width, and the channels (RGB).

The first two values are the measures of the images, and the third one tells us how the RGB values are constructed for that pixel of the images. 

```python
plt.figure(figsize=(12, 6))  # ancho=12, alto=6 pulgadas

plt.subplot(1, 2, 1)
plt.imshow(img_mountain)
plt.title("Mountain image")

plt.subplot(1, 2, 2)
plt.imshow(img_flowers)
plt.title("Flowers image")
```

```python
red_filter_flowers = img_flowers[:, :, 0]
green_filter_flowers = img_flowers[:, :, 1]
blue_filter_flowers = img_flowers[:, :, 2]
```

```python
# This code is just for represent the images
plt.figure(figsize=(12, 6))  # ancho=12, alto=6 pulgadas

rows, cols = 1, 4

plt.subplot(rows, cols, 1)
plt.imshow(img_flowers)
plt.title("Original Image")

plt.subplot(rows, cols, 2)
plt.imshow(red_filter_flowers)
plt.title("Red")

plt.subplot(rows, cols, 3)
plt.imshow(green_filter_flowers)
plt.title("Green")

plt.subplot(rows, cols, 4)
plt.imshow(blue_filter_flowers)
plt.title("Blue")
```

The history method allows viewing which values are more representative in the image; this is especially important because we can use the more important values, reducing the data to analyze and sacrificing the details that don't matter. 

```python
plt.hist(img_flowers.ravel(), bins=range(256))
```

## Manipulation

We can manipulate the image by applying different transformations to the original image, like rotations, shrinking, filters, and others.

### Normalize

Images are huge data represent in an array; when we work with images, we want to change the information to perform a model training.

```python
normalize_mountain = img_mountain / 255
normalize_mountain
```

### Shrinking and interpolation

When we shrink an image, we lose quality, and the image will be pixelated; the interpolation process applies a mathematical calculation for where some of the pixels will be to get a smoother image.

```python
shrink_mountain = img_mountain[::4, ::4] # Reduce the image 4 times 
plt.imshow(shrink_mountain)
```

```python
plt.figure(figsize=(12, 6))  # ancho=12, alto=6 pulgadas

plt.subplot(1, 2, 1)
plt.imshow(shrink_mountain, interpolation="bilinear")
plt.title("Bilinear")

plt.subplot(1, 2, 2)
plt.imshow(shrink_mountain, interpolation="bicubic")
plt.title("Bicubic")
```

### Rotations

The rotation process is fundamental. While we are working with images, we can use some methods from NumPy or the library that we are working on or apply the rotation matrix to our image given the angle that we want to rotate.

$$
\begin{bmatrix}
\cos\theta  & -\sin\theta  \\
 \sin\theta & \cos\theta \\
\end{bmatrix}
$$

```python
rotated_mountain = np.rot90(img_mountain)
plt.imshow(rotated_mountain)
```
